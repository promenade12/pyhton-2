## 字符串的乐趣
> 不做输入错误检查。
1. `is_palindrome(s)`：判断回文，忽略空格、大小写，返回布尔值。
2. `is_abecedarian(s)`：判断字母是否按非递减字母顺序，忽略空格，返回布尔。
3. `remove_vowels(s)`：移除元音字母 a,e,i,o,u，返回新字符串。


```python
def is_palindrome(s):
    # 忽略空格、全部转为小写
    processed = s.replace(" ", "").lower()
    return processed == processed[::-1]


def is_abecedarian(s):
    processed = s.replace(" ", "").lower()
    for i in range(len(processed)-1):
        if processed[i] > processed[i+1]:
            return False
    return True


def remove_vowels(s):
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    return "".join(ch for ch in s if ch not in vowels)


# 测试
print("=== is_palindrome 回文测试 ===")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))

print("\n=== is_abecedarian 字母顺序测试 ===")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))

print("\n=== remove_vowels 删除元音测试 ===")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

## 列表的乐趣
1. `list_reverse(lst)`：接收列表，返回反转后的新列表；输入不是列表抛出错误。
2. `is_sorted(seq)`：接收任意序列，判断是否升序，**只遍历一次序列**，返回布尔。


```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]


def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


# 测试
print("=== list_reverse 测试 ===")
print(list_reverse([1,2,3]))
try:
    list_reverse((1,2,3))
except TypeError as e:
    print("异常测试：", e)

print("\n=== is_sorted 测试 ===")
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
print(is_sorted((2,4,6,8)))
```

## 作为向量的元组
> 使用元组表示向量；元组的元组表示矩阵。
1. `vec_scalar_mult(t, s)`：向量标量乘法，每个元素乘标量，返回新元组；类型非法抛出`TypeError`。
2. `vec_inner_product(t1, t2)`：向量内积；非元组抛`TypeError`；长度不等抛`ValueError`。
3. `check_valid_mx(my_mx)`：校验矩阵合法性：外层全部是元组、内层元组长度全部相同、所有元素为int/float，返回布尔。
4. `mx_vec_mult(mx, vec)`：矩阵向量乘法；调用`check_valid_mx`校验输入，非法输入抛出异常，返回结果元组。

内积公式：
$$
\sum_{j=1}^{d} x_j y_j
$$


```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须为元组向量")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    if not (isinstance(t1, tuple) and isinstance(t2, tuple)):
        raise TypeError("输入必须为元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度不一致")
    return float(sum(a*b for a,b in zip(t1, t2)))


def check_valid_mx(my_mx):
    if not isinstance(my_mx, tuple):
        return False
    row_len = None
    for row in my_mx:
        if not isinstance(row, tuple):
            return False
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    row_count = len(mx)
    col_count = len(mx[0]) if row_count>0 else 0
    if len(vec) != col_count:
        raise ValueError("矩阵列数与向量维度不匹配")
    res = tuple(sum(mx[i][j] * vec[j] for j in range(col_count)) for i in range(row_count))
    return res


# 测试
print("=== vec_scalar_mult 向量标量乘 ===")
print(vec_scalar_mult((1,2,3), 2))

print("\n=== vec_inner_product 内积 ===")
print(vec_inner_product((1,2),(3,4)))

print("\n=== check_valid_mx 矩阵校验 ===")
mx_ok = ((1,2),(3,4))
mx_bad1 = ((1,2),[3,4])
mx_bad2 = ((1,),(2,3))
print(check_valid_mx(mx_ok))
print(check_valid_mx(mx_bad1))
print(check_valid_mx(mx_bad2))

print("\n=== mx_vec_mult 矩阵向量乘法 ===")
print(mx_vec_mult(mx_ok, (5,6)))
```

## 更多向量的乐趣（面向对象 Vector 类）
任务：
1. 定义`Vector`类；接收维度、元素序列；仅传维度提供默认值；维度负数/类型错误、维度与元素数量不匹配抛出异常。
2. 思考题：维度和元素应当为实例属性，每个向量对象拥有自己的维度与数据；若作为类属性，全部实例会共享同一套数据，不符合向量对象独立的语义。
3. 实现比较运算符 `== < <= >`；字典序逐坐标比较；维度不同抛出异常。
4. `dot(other)`实例方法：计算两向量内积；参数非法、维度不一致抛异常。
5. 实现乘法运算符：
    - 标量左乘、右乘（`2*v`、`v*2`）返回新Vector；
    - `v * w`：逐元素对应相乘生成新Vector；维度不一致抛出异常。


```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须为整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim
        if elements is None:
            self.elements = tuple([0]*dim)
        else:
            ele_tuple = tuple(elements)
            if len(ele_tuple) != dim:
                raise ValueError(f"维度{dim}与元素数量{len(ele_tuple)}不匹配")
            self.elements = ele_tuple

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较不同维度向量")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较不同维度向量")
        return self.elements < other.elements

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not (self <= other)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，无法计算内积")
        return sum(a*b for a,b in zip(self.elements, other.elements))

    def __mul__(self, other):
        # v * other
        if isinstance(other, (int, float)):
            new_ele = [x*other for x in self.elements]
            return Vector(self.dim, new_ele)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，无法逐元素相乘")
            new_ele = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_ele)
        else:
            raise TypeError("不支持该类型乘法")

    def __rmul__(self, scalar):
        # 标量左乘 2 * v
        return self * scalar

    def __repr__(self):
        return f"Vector(dim={self.dim}, elements={self.elements})"


# --------测试--------
print("=== Vector类测试 ===")
v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
v_zero = Vector(3)
print(v_zero)
print(v1, v2)
print("v1 < v2:", v1 < v2)
print("v1 == v2:", v1 == v2)

v3 = Vector(2, (1,2))
v4 = Vector(2, (3,4))
print("dot内积 v3.dot(v4):", v3.dot(v4))

print("标量右乘 v3*2 :", v3*2)
print("标量左乘 2*v3 :", 2*v3)
print("向量逐元素相乘 v3*v4 :", v3*v4)

# 不同维度比较会抛异常
try:
    va = Vector(2, (1,2))
    vb = Vector(3, (1,2,3))
    print(va < vb)
except ValueError as e:
    print("异常捕获：", e)
```
